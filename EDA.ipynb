```python
for month, url in urls.items():
    file_path = f"raw_data/dublinbikes_{month}.csv"
    response = requests.get(url)

    if response.status_code == 200:
        with open(file_path, "wb") as f:
            f.write(response.content)
        print(f"Downloaded: {file_path}")
    else:
        print(f"Failed to download {month}")
```

Merge all CSVs into one history file

```python
import pandas as pd
import glob

# Read all CSV files
csv_files = glob.glob("raw_data/*.csv")

df_list = []
for file in csv_files:
    df = pd.read_csv(file)
    df_list.append(df) #[]

# Combine into single DataFrame
history_df = pd.concat(df_list, ignore_index=True)

# Save final merged file
history_df.to_csv("dublinbikes_history.csv", index=False)

print("Merged file saved as dublinbikes_history.csv")
```
